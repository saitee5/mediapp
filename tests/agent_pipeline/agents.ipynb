## LangGraph StateGraph Setup
This notebook demonstrates a basic setup for a StateGraph using LangGraph.

```python
from typing import TypedDict, Annotated
import operator
from langgraph.graph import StateGraph, START, END

# Define the State
class AgentState(TypedDict):
    messages: Annotated[list, operator.add]
    # Add other state variables here
```

```python
def my_node(state: AgentState):
    print("Executing my_node")
    # Return a dictionary with the state updates
    return {"messages": ["Hello from my_node!"]}

# Initialize the StateGraph
workflow = StateGraph(AgentState)

# Add nodes to the graph
workflow.add_node("my_node", my_node)

# Set the entry point
workflow.add_edge(START, "my_node")

# Set the exit point
workflow.add_edge("my_node", END)

# Compile the graph
app = workflow.compile()
```

```python
# Test the graph
initial_state = {"messages": []}
result = app.invoke(initial_state)

print("\nFinal State:")
print(result)
```
