# Clinical Concept Extraction Agent
This notebook demonstrates an agent that takes a medical conversation transcript and uses Gemini 1.5 Flash (via Google Cloud Vertex AI) to extract and classify medical entities: symptoms, family/personal history, subjective complaints, objective findings, and preliminary diagnoses.


```
from pydantic import BaseModel, Field
from typing import List, Optional
from langchain_google_vertexai import ChatVertexAI
from langchain_core.prompts import ChatPromptTemplate
from dotenv import load_dotenv
import os

load_dotenv()
```

```
class ClinicalConcepts(BaseModel):
    symptoms: List[str] = Field(description="List of symptoms reported by the patient.")
    family_personal_history: List[str] = Field(description="Details about family or personal medical history.")
    subjective_complaints: List[str] = Field(description="Subjective complaints from the patient's perspective.")
    objective_findings: List[str] = Field(description="Objective findings observed by the clinician.")
    preliminary_diagnoses: List[str] = Field(description="Preliminary diagnoses or potential medical conditions.")
```

```
# Initialize Gemini 1.5 Flash via Vertex AI
llm = ChatVertexAI(
    model="gemini-1.5-flash",
    temperature=0,
)

# Bind the Pydantic model to the LLM to enforce structured output
structured_llm = llm.with_structured_output(ClinicalConcepts)

# Define the prompt template
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an expert medical AI assistant. Your task is to analyze clinical conversation transcripts and extract specific medical entities accurately."),
    ("human", "Please analyze the following conversation transcript and extract the clinical concepts:\n\n{transcript}")
])

# Create the extraction chain
extraction_chain = prompt | structured_llm
```

```
# Load the transcript from the file
transcript_path = "dummy_transcript_01.txt"

try:
    with open(transcript_path, 'r', encoding='utf-8') as f:
        sample_transcript = f.read()
    print(f"Successfully loaded transcript from {transcript_path}")
except FileNotFoundError:
    print(f"Transcript file {transcript_path} not found. Using fallback transcript.")
    sample_transcript = """
    Doctor: Good morning, what brings you in today?
    Patient: Hi Dr. Smith. I've been having this persistent headache for the past 3 days. It's a throbbing pain on the right side of my head.
    Doctor: I see. Any other symptoms? Nausea, sensitivity to light?
    Patient: Yes, definitely some sensitivity to light, and I feel a bit nauseous in the mornings.
    Doctor: Okay. Has anyone in your family had migraines?
    Patient: Yes, my mother suffers from severe migraines. Also, I have a history of high blood pressure, though it's been under control lately.
    Doctor: Let's check your vitals. Your blood pressure is 135/85, which is slightly elevated. Heart rate is 78. Pupils are equal and reactive.
    Patient: It just makes it really hard to focus on work.
    Doctor: Based on what you're telling me and the findings, it looks like you might be experiencing a classic migraine, possibly tension-related. We'll start you on some medication and monitor it.
    """

print("Extracting concepts...")
result = extraction_chain.invoke({"transcript": sample_transcript})

print("\n--- Extracted Clinical Concepts ---")
print(f"Symptoms: {result.symptoms}")
print(f"Family/Personal History: {result.family_personal_history}")
print(f"Subjective Complaints: {result.subjective_complaints}")
print(f"Objective Findings: {result.objective_findings}")
print(f"Preliminary Diagnoses: {result.preliminary_diagnoses}")
```
